```python
# ┌────────────────────────────────────────────────────────────────────────┐
# │  Cell 1 · Environment Setup (Shift+Enter to run)                       │
# │                                                                        │
# │  Welcome\! This notebook runs GEMC directly in your browser.            │
# │  How to use:                                                           │
# │                                                                        │
# │  1. Click any grey code cell to select it                              │
# │  2. Press `Shift + Enter` to run it (or click Run in the toolbar)      │
# │  3. Run cells in order, top to bottom                                  │
# │  4. Wait for `In [*]` to become `In [1]` before running the next cell  │
# │                                                                        │
# │  Optional cells are provided to edit the code and the YAML files.      │
# │  After editing, re-run the other cells to see the changes.             │
# │                                                                        │
# │  Example: scintillator_barrel -> documentation:                        │
# │  https://gemc.github.io/home/examples/basic/scintillator_barrel.       │
# │                                                                        │
# │  This example arranges 48 trapezoidal scintillator paddles in a        │
# │  barrel and uses flux digitization to record particle hits per         │
# │  paddle. Three 500 MeV protons are fired isotropically in phi.         │
# │                                                                        │
# │  Import notebook helpers                                               │
# │  Copy basic/scintillator_barrel example to local directory             │
# └────────────────────────────────────────────────────────────────────────┘

import subprocess
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))
from notebook_tools import edit, setup_example, run_gemc_display

from pygemc.api.run_geometry import run_geometry
setup_example("examples/basic/scintillator_barrel")
```

```python
# ┌────────────────────────────────────────────────────────────────────────┐
# │  Cell 2 · Build scintillator_barrel detector (Shift+Enter to run)      │
# │                                                                        │
# │  The geometry is rendered by PyVista in Jupyter.                       │
# │  Use the mouse to zoom/rotate/shift the view.                          │
# │  You can edit this file later; see cells below.                        │
# └────────────────────────────────────────────────────────────────────────┘

print(Path("scintillator_barrel.py").read_text())
run_geometry("scintillator_barrel.py")
```

```python
# ┌──────────────────────────────────────────────────────────┐
# │  Cell 3 · Run 10 events in GEMC (Shift+Enter to run)     │
# │                                                          │
# │  - Load scintillator_barrel gsystem                      │
# │  - Write CSV, ROOT outputs                               │
# │  - Display generated screenshot if present               │
# │                                                          │
# │  If GEMC crashes, just re-run this cell.                 │
# └──────────────────────────────────────────────────────────┘

yaml='scintillator_barrel.yaml'
driver='-g4view=[{driver: TOOLSSG_OFFSCREEN}]'
nevents="-n=10"
nthreads="-nthreads=1"

result = subprocess.run(["gemc", yaml, driver, nevents, nthreads], capture_output=True, text=True)
run_gemc_display(result)
```

```python
# ┌───────────────────────────────────────────────────────────────────────┐
# │  Cell 4 · Plot totEdep (Shift+Enter to run)                           │
# │                                                                       │
# │  Use the analyzer API on the digitized CSV file with a log-scale      │
# │  50-bin histogram.                                                    │
# └───────────────────────────────────────────────────────────────────────┘

from pygemc import read_output, plot_variable
plot_variable(read_output("barrel_t0_digitized.csv"), "totEdep", data="digitized", bins=50, logy=True, show=True)
```

```python
# ┌───────────────────────────────────────────────────────────────────────┐
# │  Cell 5 · Plot paddle identifier (Shift+Enter to run)                 │
# │                                                                       │
# │  Use the analyzer API on the digitized CSV file with a linear         │
# │  50-bin histogram to see which paddles were hit.                      │
# └───────────────────────────────────────────────────────────────────────┘

from pygemc import read_output, plot_variable
plot_variable(read_output("barrel_t0_digitized.csv"), "paddle", data="digitized", bins=50, logy=False, show=True)
```
