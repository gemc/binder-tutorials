```python
# ┌────────────────────────────────────────────────────────────────────────┐
# │  Cell 1 · Environment Setup (Shift+Enter to run)                       │
# │                                                                        │
# │  Welcome! This notebook runs GEMC directly in your browser.            │
# │  How to use:                                                           │
# │                                                                        │
# │  1. Click any grey code cell to select it                              │
# │  2. Press `Shift + Enter` to run it (or click Run in the toolbar)      │
# │  3. Run cells in order, top to bottom                                  │
# │  4. Wait for `In [*]` to become `In [1]` before running the next cell  │
# │                                                                        │
# │  Optional cells are provided to edit the code and the YAML files.      │
# │  After editing, re-run the other cells to see the changes.             │
# │                                                                        │
# │  Example: boolean_solids -> documentation:                             │
# │  https://gemc.github.io/home/examples/basic/boolean_solids.            │
# │                                                                        │
# │  This example demonstrates subtraction, union, and intersection        │
# │  boolean operations using a plate, a cross, and a lens.                │
# │                                                                        │
# │  Import notebook helpers                                               │
# │  Copy basic/boolean_solids example to local directory                  │
# └────────────────────────────────────────────────────────────────────────┘

import subprocess
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))
from notebook_tools import edit, setup_example, run_gemc_display

from pygemc.api.run_geometry import run_geometry
setup_example("examples/basic/boolean_solids")
```

```python
# ┌────────────────────────────────────────────────────────────────────────┐
# │  Cell 2 · Build boolean_solids detector (Shift+Enter to run)          │
# │                                                                        │
# │  The geometry is rendered by PyVista in Jupyter.                       │
# │  Use the mouse to zoom, rotate, or shift the view.                     │
# │  You can edit this file later; see cells below.                        │
# └────────────────────────────────────────────────────────────────────────┘

print(Path("boolean_solids.py").read_text())
run_geometry("boolean_solids.py")
```

```python
# ┌────────────────────────────────────────────────────────────────────────┐
# │  Cell 3 · Run 10 events in GEMC (Shift+Enter to run)                  │
# │                                                                        │
# │  Load boolean_solids gsystem                                           │
# │  Display generated screenshot if present                               │
# │                                                                        │
# │  If GEMC crashes, just re-run this cell.                               │
# └────────────────────────────────────────────────────────────────────────┘

yaml = "boolean_solids.yaml"
driver = '-g4view=[{driver: TOOLSSG_OFFSCREEN}]'
nevents = "-n=10"
nthreads = "-nthreads=1"
result = subprocess.run(
    ["gemc", yaml, driver, nevents, nthreads],
    capture_output=True,
    text=True,
)
run_gemc_display(result)
```

```python
# ┌────────────────────────────────────────────────────────────────────────┐
# │  Optional: Cell 4 · Edit boolean_solids.py (Shift+Enter to run)       │
# └────────────────────────────────────────────────────────────────────────┘

edit("boolean_solids.py")
```
