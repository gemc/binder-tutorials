```python
# ┌────────────────────────────────────────────────────────────────────────┐
# │  Cell 1 · Environment Setup (Shift+Enter to run)                       │
# │  GEMC in Binder                                                        │
# │                                                                        │
# │  Welcome! This notebook runs GEMC directly in your browser.            │
# │  How to use:                                                           │
# │                                                                        │
# │  1. Click any grey code cell to select it                              │
# │  2. Press `Shift + Enter` to run it (or click Run in the toolbar)      │
# │  3. Run cells in order, top to bottom                                  │
# │  4. Wait for `In [*]` to become `In [1]` before running the next cell  │
# │                                                                        │
# │  Some optional cells are provided to edit the code and the YAML        │
# │  files. After editing,                                                 │
# │  re-run the other cells to see the changes.                            │
# │                                                                        │
# │  Example: b2 -> documentation:                                         │
# │  https://gemc.github.io/home/examples/basic/b2.                        │
# │                                                                        │
# │  This example duplicates the geant4 basic example B2. It uses the      │
# │  flux digitization to save true and digitized information in the       │
# │  output.                                                               │
# │                                                                        │
# │                                                                        │
# │  Import notebook helpers                                               │
# │  Copy basic/b2 example to local directory                              │
# └────────────────────────────────────────────────────────────────────────┘

import subprocess
import sys
from pathlib import Path
from IPython.display import Image, display

sys.path.insert(0, str(Path.cwd().parent))
from notebook_tools import edit, setup_example

from pygemc.api.run_geometry import run_geometry
setup_example("examples/basic/b2")
```

```python
# ┌─────────────────────────────────────────────────────┐
# │  Cell 3 · Build b2 detector (Shift+Enter to run)    │
# │                                                     │
# │  The geometry is rendered by PyVista in Jupyter.    │
# │  Use the mouse to zoom/rotate/shift the view.       │
# └─────────────────────────────────────────────────────┘

run_geometry("b2.py")
```

```python
# ┌─────────────────────────────────────────────────────────┐
# │  Cell 4 · Run 1000 events in GEMC (Shift+Enter to run)  │
# │                                                         │
# │  - Load b2 gsystem                                      │
# │  - Write CSV, ROOT outputs                              │
# │  - Display generated screenshot if present              │
# └─────────────────────────────────────────────────────────┘

yaml='b2.yaml'
driver='-g4view=[{driver: TOOLSSG_OFFSCREEN}]'
camera='-g4camera=[{phi: -10*deg, theta: 250*deg}]'
light='-g4light=[{phi: 160*deg, theta: 120*deg}]'
nevents="-n=50"
nthreads="-nthreads=1"

result = subprocess.run(["gemc", yaml, driver, camera, light, nevents, nthreads], capture_output=True, text=True)
if result.returncode != 0:
    print("GEMC failed:")
    print(result.stderr)
else:
    print(result.stdout)

image_file = Path("gemc_run_0.png")
if image_file.exists():
    display(Image(filename=str(image_file)))
else:
    print(f"No GEMC image found: {image_file}")

```

```python
# ┌───────────────────────────────────────────────────────────────────────┐
# │  Cell 6 · Plot totEdep (Shift+Enter to run)                           │
# │                                                                       │
# │  Use the analyzer API on the digitized CSV file with a linear 50-bin  │
# │  histogram.                                                           │
# └───────────────────────────────────────────────────────────────────────┘

from pygemc import read_output, plot_variable
plot_variable(read_output("CO2_t0_digitized.csv"), "totEdep", data="digitized", bins=50, logy=True, show=True)
```

```python
# ┌───────────────────────────────────────────────────────────────────────┐
# │  Cell 7 · Plot true-info vz (Shift+Enter to run)                      │
# │                                                                       │
# │  Use the analyzer API on the true-info CSV file with a linear 50-bin  │
# │  histogram.                                                           │
# └───────────────────────────────────────────────────────────────────────┘

from pygemc import read_output, plot_variable
plot_variable(read_output("CO2_t0_true_info.csv"), "vz", data="true_info", bins=50, logy=False, show=True)
```
