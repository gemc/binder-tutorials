```python
# ┌────────────────────────────────────────────────────────────────────────┐
# │  Cell 1 · Environment Setup (Shift+Enter to run)                       │
# │  GEMC in Binder                                                        │
# │                                                                        │
# │  Welcome! This notebook runs GEMC directly in your browser.            │
# │  How to use:                                                           │
# │                                                                        │
# │  1. Click any grey code cell to select it                              │
# │  2. Press `Shift + Enter` to run it (or click Run in the toolbar)      │
# │  3. Run cells in order, top to bottom                                  │
# │  4. Wait for `In [*]` to become `In [1]` before running the next cell  │
# │                                                                        │
# │  Some optional cells are provided to edit the code and the YAML        │
# │  files. After editing,                                                 │
# │  re-run the other cells to see the changes.                            │
# │                                                                        │
# │  Example: simple_flux -> documentation:                                │
# │  https://gemc.github.io/home/examples/basic/simple_flux.               │
# │                                                                        │
# │  This setup uses a cylindrical liquid hydrogen target (`G4_lH2`) and   │
# │  an air flux plane (`G4_AIR`).                                         │
# │                                                                        │
# │                                                                        │
# │  import PyVista and editor                                             │
# │  Copy basic/simple_flux example to local directory                     │
# └────────────────────────────────────────────────────────────────────────┘

import subprocess, warnings, sys, vtk
import pyvista as pv
vtk.vtkObject.GlobalWarningDisplayOff()  # suppress warnings

from pygemc.api.run_geometry import run_geometry
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))
from notebook_tools import edit, setup_example
setup_example("examples/basic/simple_flux")
```

```python
# ┌───────────────────────────────────────────────────────────────────────┐
# │  Cell 3 · Build simple_flux detector (Shift+Enter to run)             │
# │                                                                       │
# │  The geometry is rendered by PyVista in Jupyter. The target material  │
# │  is liquid hydrogen (`G4_lH2`).                                       │
# │  Use the mouse to zoom, rotate, or shift the view.                    │
# │                                                                       │
# │  Edit the geometry in the step above to change the geometry - re-run  │
# │  this cell to update the display.                                     │
# └───────────────────────────────────────────────────────────────────────┘

run_geometry("simple_flux.py")
```

```python
# ┌───────────────────────────────────────────────────────────┐
# │  Cell 4 · Run 10,000 events in GEMC (Shift+Enter to run)  │
# │                                                           │
# │  - Load simple_flux gsystem                               │
# │  - Write CSV, ROOT outputs                                │
# │  - Display generated screenshot if present                │
# └───────────────────────────────────────────────────────────┘

from pathlib import Path
from IPython.display import Image, display

yaml='simple_flux.yaml'
driver='-g4view=[{driver: TOOLSSG_OFFSCREEN}]'
camera='-g4camera=[{phi: -10*deg, theta: 250*deg}]'
light='-g4light=[{phi: 160*deg, theta: 120*deg}]'
nevents="-n=100"
nthreads="-nthreads=1"

result = subprocess.run(["gemc", yaml, driver, camera, light, nevents, nthreads], capture_output=True, text=True)
if result.returncode != 0:
    print("GEMC failed:")
    print(result.stderr)
else:
    print(result.stdout)

image_file = Path("gemc_run_0.png")
if image_file.exists():
    display(Image(filename=str(image_file)))
else:
    print(f"No GEMC image found: {image_file}")
    
    
result = subprocess.run(
	["ls", "-l"],
	capture_output=True,
	text=True,
)

print(result.stdout)
```
