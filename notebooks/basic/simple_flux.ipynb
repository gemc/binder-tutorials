```python
# ┌────────────────────────────────────────────────────────────────────────┐
# │  Cell 1 · Environment setup (Shift+Enter to run)                       │
# │                                                                        │
# │  Welcome! This notebook runs GEMC directly in your browser.            │
# │  How to use:                                                           │
# │                                                                        │
# │  1. Click any grey code cell to select it                              │
# │  2. Press `Shift + Enter` to run it (or click Run in the toolbar)      │
# │  3. Run cells in order, top to bottom                                  │
# │  4. Wait for `In [*]` to change to a number before continuing          │
# │                                                                        │
# │  Optional cells are provided to edit the code and the YAML files.      │
# │  After editing, re-run the other cells to see the changes.             │
# │                                                                        │
# │  Documentation:                                                        │
# │  https://gemc.github.io/home/examples/basic/simple_flux/               │
# │                                                                        │
# │  This setup uses a cylindrical liquid hydrogen target (`G4_lH2`) and   │
# │  an air flux plane (`G4_AIR`).                                         │
# │                                                                        │
# │                                                                        │
# │  Import notebook helpers                                               │
# │  Copy basic/simple_flux example to local directory                     │
# └────────────────────────────────────────────────────────────────────────┘

import subprocess
import sys
from pathlib import Path

notebooks_dir = next(
    parent
    for parent in (Path.cwd(), *Path.cwd().parents)
    if (parent / "notebook_tools").is_dir()
)
sys.path.insert(0, str(notebooks_dir))
from notebook_tools import edit, run_gemc_display, setup_example

from pygemc.api.run_geometry import run_geometry
setup_example("examples/basic/simple_flux")
```

```python
# ┌────────────────────────────────────────────────────────────────────────┐
# │  Cell 2 · Build the simple_flux geometry (Shift+Enter to run)          │
# │                                                                        │
# │  Print simple_flux.py, write gemc.db, and render the geometry with     │
# │  PyVista. The target is liquid hydrogen (G4_lH2). Use the mouse to     │
# │  zoom, rotate, or pan the view.                                        │
# └────────────────────────────────────────────────────────────────────────┘

print(Path("simple_flux.py").read_text())
run_geometry("simple_flux.py")
```

```python
# ┌────────────────────────────────────────────────────────────────────────┐
# │  Cell 3 · Run 100 events in GEMC (Shift+Enter to run)                  │
# │                                                                        │
# │  Load the simple_flux geometry, write CSV output, and display the      │
# │  off-screen GEMC image. If GEMC fails during startup, re-run this.     │
# └────────────────────────────────────────────────────────────────────────┘

yaml = "simple_flux.yaml"
driver = '-g4view=[{driver: TOOLSSG_OFFSCREEN}]'
nevents = "-n=100"
nthreads = "-nthreads=1"
result = subprocess.run(
    ["gemc", yaml, driver, nevents, nthreads],
    capture_output=True,
    text=True,
)
run_gemc_display(result)
```

```python
# ┌────────────────────────────────────────────────────────────────────────┐
# │  Cell 4 · Plot total energy deposited (Shift+Enter to run)             │
# │                                                                        │
# │  Use the analyzer API on the digitized CSV file with a log-scale       │
# │  50-bin histogram.                                                     │
# └────────────────────────────────────────────────────────────────────────┘

from pygemc import plot_variable, read_output

plot_variable(
    read_output("simple_flux_t0_digitized.csv"),
    "totEdep",
    bins=50,
    logy=True,
    show=True,
)
```

```python
# ┌────────────────────────────────────────────────────────────────────────┐
# │  Cell 5 · Plot time (Shift+Enter to run)                               │
# │                                                                        │
# │  Use the analyzer API on the digitized CSV file with a log-scale       │
# │  50-bin histogram.                                                     │
# └────────────────────────────────────────────────────────────────────────┘

from pygemc import plot_variable, read_output

plot_variable(
    read_output("simple_flux_t0_digitized.csv"),
    "time",
    bins=50,
    logy=True,
    show=True,
)
```
