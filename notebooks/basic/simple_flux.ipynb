### Cell 4 · Run 100 events in GEMC (Shift+Enter to run)

- Load simple_flux gsystem
- Write CSV, ROOT outputs
- Display generated screenshot if present


```python
from pathlib import Path
from IPython.display import Image, display

yaml='simple_flux.yaml'
driver='-g4view=[{driver: TOOLSSG_OFFSCREEN}]'
camera='-g4camera=[{phi: 0*deg, theta: 270*deg}]'
nevents="-n=100"
nthreads="-nthreads=1"

result = subprocess.run(["gemc", yaml, driver, camera, nevents, nthreads], capture_output=True, text=True)
if result.returncode != 0:
    print("GEMC failed:")
    print(result.stderr)
else:
    print(result.stdout)

image_file = Path("gemc_run_0.png")
if image_file.exists():
    display(Image(filename=str(image_file)))
else:
    print(f"No GEMC image found: {image_file}")
    
    
result = subprocess.run(
	["ls", "-l"],
	capture_output=True,
	text=True,
)

print(result.stdout)
```
