```python
# ┌────────────────────────────────────────────────────────────────────────┐
# │  Cell 1 · Environment Setup (Shift+Enter to run)                       │
# │                                                                        │
# │  Welcome! This notebook runs GEMC directly in your browser.            │
# │  How to use:                                                           │
# │                                                                        │
# │  1. Click any grey code cell to select it                              │
# │  2. Press `Shift + Enter` to run it (or click Run in the toolbar)      │
# │  3. Run cells in order, top to bottom                                  │
# │  4. Wait for `In [*]` to become `In [1]` before running the next cell  │
# │                                                                        │
# │  Optional cells are provided to edit the code and the YAML files.      │
# │  After editing, re-run the other cells to see the changes.             │
# │                                                                        │
# │  Example: materials -> documentation:                                  │
# │  https://gemc.github.io/home/examples/basic/materials.                 │
# │                                                                        │
# │  This example demonstrates Geant4 built-in materials, custom           │
# │  molecular and fractional-mass materials, and optical material         │
# │  properties.                                                           │
# │                                                                        │
# │  Import notebook helpers                                               │
# │  Copy basic/material example to local directory                        │
# └────────────────────────────────────────────────────────────────────────┘

import subprocess
import sys
from pathlib import Path
from IPython.display import Image, display

sys.path.insert(0, str(Path.cwd().parent))
from notebook_tools import edit, setup_example

from pygemc.api.run_geometry import run_geometry
setup_example("examples/basic/material")
```

```python
# ┌──────────────────────────────────────────────────────────┐
# │  Cell 2 · Build materials detector (Shift+Enter to run)  │
# │                                                          │
# │  The geometry is rendered by PyVista in Jupyter.         │
# │  Use the mouse to zoom/rotate/shift the view.            │
# │  You can edit this file later; see cells below.          │
# └──────────────────────────────────────────────────────────┘

print(Path("materials.py").read_text())
run_geometry("materials.py")
```

```python
# ┌────────────────────────────────────────────────────────┐
# │  Cell 3 · Run 1 event in GEMC (Shift+Enter to run)     │
# │                                                        │
# │  - Load materials gsystem                              │
# │  - Write CSV, ROOT outputs                             │
# │  - Display generated screenshot if present             │
# │                                                        │
# │  If GEMC crashes, just re-run this cell.               │
# └────────────────────────────────────────────────────────┘

yaml='materials.yaml'
driver='-g4view=[{driver: TOOLSSG_OFFSCREEN}]'
nevents="-n=1"
nthreads="-nthreads=1"
result = subprocess.run(["gemc", yaml, driver, nevents, nthreads], capture_output=True, text=True)
if result.returncode != 0:
    print("GEMC failed:")
    print(result.stderr)
    print("If this was an intermittent startup or display failure, re-run this cell.")
else:
    print(result.stdout)

image_file = Path("gemc_run_0.png")
if image_file.exists():
    display(Image(filename=str(image_file)))
else:
    print(f"No GEMC image found: {image_file}")

```

```python
# ┌─────────────────────────────────────────────────────────────┐
# │  Optional: Cell 4 · Edit materials.py (Shift+Enter to run)  │
# └─────────────────────────────────────────────────────────────┘

edit("materials.py")
```
