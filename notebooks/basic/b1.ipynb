```python
# ┌────────────────────────────────────────────────────────────────────────┐
# │  Cell 1 · Environment Setup (Shift+Enter to run)                       │
# │                                                                        │
# │  Welcome! This notebook runs GEMC directly in your browser.            │
# │  How to use:                                                           │
# │                                                                        │
# │  1. Click any grey code cell to select it                              │
# │  2. Press `Shift + Enter` to run it (or click Run in the toolbar)      │
# │  3. Run cells in order, top to bottom                                  │
# │  4. Wait for `In [*]` to become `In [1]` before running the next cell  │
# │                                                                        │
# │  Optional cells are provided to edit the code and the YAML files.      │
# │  After editing, re-run the other cells to see the changes.             │
# │                                                                        │
# │  Example: b1 -> documentation:                                         │
# │  https://gemc.github.io/home/examples/basic/b1.                        │
# │                                                                        │
# │  This example duplicates the geant4 basic example B1. It uses the      │
# │  dosimeter digitization to store the accumulated dose on a detector.   │
# │                                                                        │
# │                                                                        │
# │  Import notebook helpers                                               │
# │  Copy basic/b1 example to local directory                              │
# └────────────────────────────────────────────────────────────────────────┘

import subprocess
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))
from notebook_tools import edit, setup_example, run_gemc_display

from pygemc.api.run_geometry import run_geometry
setup_example("examples/basic/b1")
```

```python
# ┌─────────────────────────────────────────────────────┐
# │  Cell 2 · Build b1 detector (Shift+Enter to run)    │
# │                                                     │
# │  The geometry is rendered by PyVista in Jupyter.    │
# │  Use the mouse to zoom/rotate/shift the view.       │
# │  You can edit this file later; see cells below.     │
# └─────────────────────────────────────────────────────┘

print(Path("b1.py").read_text())
run_geometry("b1.py")
```

```python
# ┌─────────────────────────────────────────────────────────┐
# │  Cell 3 · Run 100 events in GEMC (Shift+Enter to run)   │
# │                                                         │
# │  - Load b1 gsystem                                      │
# │  - Write CSV, ROOT outputs                              │
# │  - Display generated screenshot if present              │
# │                                                         │
# │  If GEMC crashes, just re-run this cell.                │
# └─────────────────────────────────────────────────────────┘

yaml='b1.yaml'
driver='-g4view=[{driver: TOOLSSG_OFFSCREEN}]'
nevents="-n=100"
nthreads="-nthreads=1"
result = subprocess.run(["gemc", yaml, driver, nevents, nthreads], capture_output=True, text=True)
run_gemc_display(result)
```

```python
# ┌───────────────────────────────────────────────────────────────────────┐
# │  Cell 4 · Plot dose (Shift+Enter to run)                              │
# │                                                                       │
# │  Use the analyzer API on the digitized CSV file with a linear 50-bin  │
# │  histogram.                                                           │
# └───────────────────────────────────────────────────────────────────────┘

from pygemc import read_output, plot_variable
plot_variable(read_output("b1_t0_digitized.csv"), "dose", data="digitized", bins=50, logy=False, show=True)
```

```python
# ┌───────────────────────────────────────────────────────────────────────┐
# │  Cell 5 · Plot etot (Shift+Enter to run)                              │
# │                                                                       │
# │  Use the analyzer API on the digitized CSV file with a linear 50-bin  │
# │  histogram.                                                           │
# └───────────────────────────────────────────────────────────────────────┘

from pygemc import read_output, plot_variable
plot_variable(read_output("b1_t0_digitized.csv"), "etot", data="digitized", bins=50, logy=True, show=True)
```
