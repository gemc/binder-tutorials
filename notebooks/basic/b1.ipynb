```python
# ┌────────────────────────────────────────────────────────────────────────┐
# │  Cell 1 · Environment setup (Shift+Enter to run)                       │
# │                                                                        │
# │  Welcome! This notebook runs GEMC directly in your browser.            │
# │  How to use:                                                           │
# │                                                                        │
# │  1. Click any grey code cell to select it                              │
# │  2. Press `Shift + Enter` to run it (or click Run in the toolbar)      │
# │  3. Run cells in order, top to bottom                                  │
# │  4. Wait for `In [*]` to change to a number before continuing          │
# │                                                                        │
# │  Optional cells are provided to edit the code and the YAML files.      │
# │  After editing, re-run the other cells to see the changes.             │
# │                                                                        │
# │  Documentation:                                                        │
# │  https://gemc.github.io/home/examples/basic/b1/                        │
# │                                                                        │
# │  This example reproduces the Geant4 basic example B1. It uses the      │
# │  dosimeter digitization to store the accumulated dose on a detector.   │
# │                                                                        │
# │                                                                        │
# │  Import notebook helpers                                               │
# │  Copy basic/b1 example to local directory                              │
# └────────────────────────────────────────────────────────────────────────┘

import subprocess
import sys
from pathlib import Path

notebooks_dir = next(
    parent
    for parent in (Path.cwd(), *Path.cwd().parents)
    if (parent / "notebook_tools").is_dir()
)
sys.path.insert(0, str(notebooks_dir))
from notebook_tools import edit, run_gemc_display, setup_example

from pygemc.api.run_geometry import run_geometry
setup_example("examples/basic/b1")
```

```python
# ┌────────────────────────────────────────────────────────────────────────┐
# │  Cell 2 · Build the b1 geometry (Shift+Enter to run)                   │
# │                                                                        │
# │  Print b1.py, write gemc.db, and render the geometry with PyVista.     │
# │  Use the mouse to zoom, rotate, or pan the view.                       │
# └────────────────────────────────────────────────────────────────────────┘

print(Path("b1.py").read_text())
run_geometry("b1.py")
```

```python
# ┌────────────────────────────────────────────────────────────────────────┐
# │  Cell 3 · Run 100 events in GEMC (Shift+Enter to run)                  │
# │                                                                        │
# │  Load the b1 geometry, write CSV output, and display the off-screen    │
# │  GEMC image. If GEMC fails during startup, re-run this cell.           │
# └────────────────────────────────────────────────────────────────────────┘

yaml = "b1.yaml"
driver = '-g4view=[{driver: TOOLSSG_OFFSCREEN}]'
nevents = "-n=100"
nthreads = "-nthreads=1"
result = subprocess.run(
    ["gemc", yaml, driver, nevents, nthreads],
    capture_output=True,
    text=True,
)
run_gemc_display(result)
```

```python
# ┌────────────────────────────────────────────────────────────────────────┐
# │  Cell 4 · Plot accumulated dose (Shift+Enter to run)                   │
# │                                                                        │
# │  Use the analyzer API on the digitized CSV file with a linear 50-bin   │
# │  histogram.                                                            │
# └────────────────────────────────────────────────────────────────────────┘

from pygemc import plot_variable, read_output

plot_variable(
    read_output("b1_t0_digitized.csv"),
    "dose",
    bins=50,
    logy=False,
    show=True,
)
```

```python
# ┌────────────────────────────────────────────────────────────────────────┐
# │  Cell 5 · Plot total deposited energy (Shift+Enter to run)             │
# │                                                                        │
# │  Use the analyzer API on the digitized CSV file with a log-scale       │
# │  50-bin histogram of etot.                                             │
# └────────────────────────────────────────────────────────────────────────┘

from pygemc import plot_variable, read_output

plot_variable(
    read_output("b1_t0_digitized.csv"),
    "etot",
    bins=50,
    logy=True,
    show=True,
)
```
