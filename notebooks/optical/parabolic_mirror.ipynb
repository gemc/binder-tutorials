```python
# ┌────────────────────────────────────────────────────────────────────────┐
# │  Cell 1 · Environment Setup (Shift+Enter to run)                       │
# │                                                                        │
# │  Welcome! This notebook runs GEMC directly in your browser.            │
# │  How to use:                                                           │
# │                                                                        │
# │  1. Click any grey code cell to select it                              │
# │  2. Press `Shift + Enter` to run it (or click Run in the toolbar)      │
# │  3. Run cells in order, top to bottom                                  │
# │  4. Wait for `In [*]` to become `In [1]` before running the next cell  │
# │                                                                        │
# │  Optional cells are provided to edit the code and the YAML files.      │
# │  After editing, re-run the other cells to see the changes.             │
# │                                                                        │
# │  Example: parabolic_mirror -> documentation:                           │
# │  https://gemc.github.io/home/examples/optical/parabolic_mirror.        │
# │                                                                        │
# │  This example sends a parallel optical-photon bundle onto a thin        │
# │  parabolic mirror shell and counts photons near the nominal focus.      │
# │                                                                        │
# │  Import notebook helpers                                               │
# │  Copy optical/parabolic_mirror example to local directory              │
# └────────────────────────────────────────────────────────────────────────┘

import subprocess
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))
from notebook_tools import edit, setup_example, run_gemc_display

from pygemc.api.run_geometry import run_geometry
setup_example("examples/optical/parabolic_mirror")
```

```python
# ┌────────────────────────────────────────────────────────────────────────┐
# │  Cell 2 · Build parabolic_mirror detector (Shift+Enter to run)         │
# │                                                                        │
# │  The geometry is rendered by PyVista in Jupyter.                       │
# │  Use the mouse to zoom, rotate, or shift the view.                     │
# │  You can edit this file later; see cells below.                        │
# └────────────────────────────────────────────────────────────────────────┘

print(Path("parabolic_mirror.py").read_text())
run_geometry("parabolic_mirror.py")
```

```python
# ┌────────────────────────────────────────────────────────────────────────┐
# │  Cell 3 · Run 1 event in GEMC (Shift+Enter to run)                     │
# │                                                                        │
# │  Load parabolic_mirror gsystem                                         │
# │  Write CSV, ROOT outputs                                               │
# │  Display generated screenshot if present                               │
# │                                                                        │
# │  If GEMC crashes, just re-run this cell.                               │
# └────────────────────────────────────────────────────────────────────────┘

yaml = "parabolic_mirror.yaml"
driver = '-g4view=[{driver: TOOLSSG_OFFSCREEN}]'
nevents = "-n=1"
nthreads = "-nthreads=1"
result = subprocess.run(
    ["gemc", yaml, driver, nevents, nthreads],
    capture_output=True,
    text=True,
)
run_gemc_display(result)
```

```python
# ┌────────────────────────────────────────────────────────────────────────┐
# │  Cell 4 · Plot photon energy (Shift+Enter to run)                      │
# │                                                                        │
# │  Use the analyzer API on the digitized CSV file with a log-scale       │
# │  50-bin histogram.                                                     │
# └────────────────────────────────────────────────────────────────────────┘

from pygemc import read_output, plot_variable
plot_variable(
    read_output("parabolic_mirror_t0_digitized.csv"),
    "E",
    bins=50,
    logy=True,
    show=True,
)
```

```python
# ┌────────────────────────────────────────────────────────────────────────┐
# │  Cell 5 · Plot panel identifier (Shift+Enter to run)                   │
# │                                                                        │
# │  Use the analyzer API on the digitized CSV file with a linear 50-bin   │
# │  histogram.                                                            │
# │  Confirm photons reached the focus detector.                           │
# └────────────────────────────────────────────────────────────────────────┘

from pygemc import read_output, plot_variable
plot_variable(
    read_output("parabolic_mirror_t0_digitized.csv"),
    "panel",
    bins=50,
    logy=False,
    show=True,
)
```
