This example showcases the activation of the Cherenkov radiation in a GEMC detector.

<br/>


```python
# ┌─────────────────────────────────────────────────────┐
# │  Cell 1 · Environment Setup (Shift+Enter to run)    │
# │  import PyVista and editor                          │
# │  Copy optical/cherenkov example to local directory  │
# └─────────────────────────────────────────────────────┘

import subprocess, warnings, sys, vtk
import pyvista as pv
vtk.vtkObject.GlobalWarningDisplayOff()  # suppress warnings

from run_geometry import run_geometry
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))
from notebook_tools import edit, setup_example
setup_example("examples/optical/cherenkov")

```

```python
# ┌─────────────────────────────────────────────────────────────┐
# │  Optional: Cell 2 · Edit cherenkov.py (Shift+Enter to run)  │
# └─────────────────────────────────────────────────────────────┘

edit("cherenkov.py")
```

```python
# ┌──────────────────────────────────────────────────────────┐
# │  Cell 3 · Build cherenkov detector (Shift+Enter to run)  │
# │                                                          │
# │  The geometry is rendered by PyVista in Jupyter.         │
# │  Use the mouse to zoom/rotatoe/shift the view.           │
# └──────────────────────────────────────────────────────────┘

run_geometry("cherenkov.py")
```

```python
# ┌────────────────────────────────────────────────────────────┐
# │  Cell 4 · Run 10 events in GEMC (Shift+Enter to run)       │
# │                                                            │
# │  - Load cherenkov gsystem                                  │
# │  - Write CSV, ROOT outputs                                 │
# │  - Display generated screenshot if present                 │
# └────────────────────────────────────────────────────────────┘

from pathlib import Path
from IPython.display import Image, display

yaml='cherenkov.yaml'
driver='-g4view=[{driver: TOOLSSG_OFFSCREEN}]'
camera='-g4camera=[{phi: -10*deg, theta: 250*deg}]'
nevents="-n=10"
nthreads="-nthreads=1"

result = subprocess.run(["gemc", yaml, driver, camera, nevents, nthreads], capture_output=True, text=True)
if result.returncode != 0:
    print("GEMC failed:")
    print(result.stderr)
else:
    print(result.stdout)

image_file = Path("gemc_run_0.png")
if image_file.exists():
    display(Image(filename=str(image_file)))
else:
    print(f"No GEMC image found: {image_file}")
    
    
result = subprocess.run(
	["ls", "-l"],
	capture_output=True,
	text=True,
)

print(result.stdout)
```
